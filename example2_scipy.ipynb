```python
from scipy.optimize import linprog
```

Example 2 from: https://realpython.com/linear-programming-python/

Maximize:  
20x_1 + 12x_2 + 40x_3 + 25x_4  

Subject to:  
x_1 + x_2 + x_3 + x_4 <= 50  
3x_1 + 2x_2 + x_3 <= 100  
x_2 + 2x_3 + 3x_4 <= 90  
x_1, x_2, x_3, x_4 >= 0  

I'm trying to do this one largely by myself rather than copying the example form. So this means my variable names are more my own.

```python
objective_function = [-20, -12, -40, -25]

constraints_lhs = [
    [1, 1, 1, 1],
    [3, 2, 1, 0],
    [0, 1, 2, 3]
]

constraints_rhs = [50, 100, 90]

positive = (0, float("inf"))
bounds = [positive, positive, positive, positive]

opt = linprog(c=objective_function, A_ub=constraints_lhs, b_ub=constraints_rhs, bounds=bounds, method="revised simplex")
opt
```
