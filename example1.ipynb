```python
from scipy.optimize import linprog
```

This is for the following problem taken from Example 1 from https://realpython.com/linear-programming-python/

Maximize:  
z = x + 2y (multipled by -1 to turn it into a minimize formula)

Subject to:  
2x + y <= 20  
-4x + 5y <= 10  
-x + 2y >= 2 (multipled by -1 to turn it into <= form)  
-x + 5 = 15
x >= 0  
y >= 0


```python
obj = [-1, -2]
#      ─┬  ─┬
#       │   └┤ Coefficient for y
#       └────┤ Coefficient for x

lhs_ineq = [[ 2,  1],  # Red constraint left side
            [-4,  5],  # Blue constraint left side
            [ 1, -2]]  # Yellow constraint left side

rhs_ineq = [20,  # Red constraint right side
            10,  # Blue constraint right side
             2]  # Yellow constraint right side

lhs_eq = [[-1, 5]]  # Green constraint left side
rhs_eq = [15]       # Green constraint right side
```

```python
bnd = [(0, float("inf")),  # Bounds of x
       (0, float("inf"))]  # Bounds of y
```

We optimize by calling _linprog_, and we pass in each of the arguments as named parameters.

```python
opt = linprog(c=obj, A_ub=lhs_ineq, b_ub=rhs_ineq, A_eq=lhs_eq, b_eq=rhs_eq, bounds=bnd, method="interior-point")
opt
```

Now let's drop the equality constraint. We just need to remove teh A_eq and b_eq constraints.

```python
opt = linprog(c=obj, A_ub=lhs_ineq, b_ub=rhs_ineq, bounds=bnd, method="revised simplex")
opt
```
